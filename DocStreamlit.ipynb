### Como trabalhar com tabelas no streamlit: 

- Necessário utilizar pandas para DataFrame
    
    `pip install pandas`
    

| st.table() | st.dataframe()|
---------------| ---------------
| Exibe uma tabela estática, não permite ordernar, filtrar, focada em apresentações fixas | Ordena, destaque de linhas e células, permite filtragem, parecida com planilha interativa|
 

```python
import pandas as pd
df = pd.DataFrame({
    'Nome': ['Ana', 'Bruno', 'Carla', 'Daniel'],
    'Idade': [23, 35, 31, 29],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte', 'Curitiba']
})
st.dataframe(df) # Exibe tabela interativa
```

#### Saída: 


```python
df = pd.DataFrame({
    'Nome': ['Ana', 'Bruno', 'Carla', 'Daniel'],
    'Idade': [23, 35, 31, 29],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte', 'Curitiba']
})
st.table(df) # Exibe tabela estática
```

### Elementos de gráfico:

O Streamlit oferece suporte a diversas bibliotecas de gráficos diferentes. A biblioteca mais básica é o Matplotlib. Para gráficos 2D tem o Vega Lite e para ps gráficos 3D e mapas há o deck.gl.

- Também há os "nativos" do Streamlit:
    1. st.line_chart()

    4. st.scatter_chart()

```python
# Exemplo:
st.line_char(dados_api)
st.area_chart(dados_api)
st.bar_chart(dados_api)
st.scatter_chart(dados_api)
st.map(dados_api) # o dataframe precisa ter as informações como 
                #latitude e longitude para o gráfico funcionar
```

> Para mais informações sobre exibir os elementos gráficos acesse: https://docs.streamlit.io/develop/api-reference/charts

### Estilização do dashboard:

Diferentemente do HTML o streamlit é limitado em relação a estilização da página e das possibilidades que poderemos implementar. Porém, para integrar código HTML no streamlit há:
    `st.html(body, *, width="stretch")`

```python
# Exemplo
import streamlit as st

st.html(
    "<p><span style='text-decoration: line-through double red;'>Oops</span>!</p>"
)
```

> Para mais informações sobre st.html, acesse: https://docs.streamlit.io/develop/api-reference/text/st.html

- Continuando sobre a estilização, poderemos adicionar colunas, o que permite:
    - Inserir contêineres dispostos em colunas lado a lado.
    -Insere vários contêineres com múltiplos elementos dispostos lado a lado e retorna uma lista de objetos contêiner.
    - Para adicionar elementos aos contêineres retornados, você pode usar a notação `with` (preferencial) ou simplesmente chamar métodos diretamente no objeto retornado. Veja os exemplos abaixo.

- Assinatura da função: `st.columns(spec, *, gap="small", vertical_alignment="top", border=False, width="stretch")`

```python
# Exemplo 
import streamlit as st

col1, col2, col3 = st.columns(3)

with col1:
    st.header("A cat")
    st.image("https://static.streamlit.io/examples/cat.jpg")

with col2:
    st.header("A dog")
    st.image("https://static.streamlit.io/examples/dog.jpg")

with col3:
    st.header("An owl")
    st.image("https://static.streamlit.io/examples/owl.jpg")
```
